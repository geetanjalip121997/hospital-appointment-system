```python
registered_id = 4521
patient_balance = 450
fee_general = 400
fee_eye = 600
fee_skin = 500 

patient_id = int(input("Enter the registered patient ID = "))

if patient_id == registered_id:
    
    age = int(input("Enter the age of patient ="))
    
    if age >= 1 and age <= 120:
        
        department = int(input("Enter the department ="))
        if department == 1:
            fee = fee_general
        elif department == 2:
            fee = fee_eye
        elif department == 3:
            fee = fee_skin
        else:
            fee = 0
            print("Invalid department")
            
        if department >= 1 and department <= 3:
            if age >= 60:
                    discount = fee * 0.50
            elif age <= 12:
                    discount = fee * 0.25
            else:
                    discount = 0

            total_fee = fee - discount
            
            if total_fee <= patient_balance:
                patient_balance = patient_balance - total_fee

                print("Appointment confirmed")
                print("Fee paid: Rs.", total_fee)
                print("Remaining balance: Rs.",
                      patient_balance)
            else:
                print("Insufficient balance")

    else :
        print ("Enter the correct age")
else :
    print(" Patient is not registered")
```
